```python
c = float(input("Enter Celsius: "))
f = (c * 9/5) + 32
print("Fahrenheit = {:.2f}".format(f))
```

```python
basic = float(input("Enter Basic Salary: "))

hra = basic * 20/100
da = basic * 15/100
pf = basic * 8/100

net = basic + hra + da - pf

print("HRA =", hra)
print("DA =", da)
print("PF =", pf)
print("Net Salary =", net)
```

```python
year = int(input("Enter year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
m1 = float(input("Enter mark 1: "))
m2 = float(input("Enter mark 2: "))
m3 = float(input("Enter mark 3: "))

avg = (m1 + m2 + m3) / 3

print("Average =", avg)

if avg >= 90:
    print("Grade A+")
elif avg >= 75:
    print("Grade A")
elif avg >= 50:
    print("Grade B")
else:
    print("Fail")
```

```python
age = int(input("Enter age: "))

vote = "Eligible" if age >= 18 else "Not Eligible"
discount = "Yes" if age >= 60 else "No"

print("Voting:", vote)
print("Senior Citizen Discount:", discount)
```

```python
n = int(input("Enter number: "))

for i in range(1, 11):
    print(n, "x", i, "=", n * i)
```

```python
n = int(input("Enter N: "))

i = 1
sum = 0

while i <= n:
    sum = sum + i
    i = i + 1

average = sum / n

print("Sum =", sum)
print("Average =", average)
```

```python
n = int(input("Enter number: "))

for i in range(2, n):
    if n % i == 0:
        print("Not Prime")
        break
else:
    print("Prime")
```

```python
n = int(input("Enter number: "))

temp = n
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

print("Reverse =", rev)

if temp == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
code = int(input("Enter HTTP code: "))

match code:
    case 200:
        print("OK")
    case 400:
        print("Bad Request")
    case 404:
        print("Not Found")
    case 500:
        print("Internal Server Error")
    case _:
        print("Invalid Code")
```
